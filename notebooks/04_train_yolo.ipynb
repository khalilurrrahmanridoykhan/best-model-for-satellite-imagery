# 04 -- Train the YOLO

Trains a YOLO11n (Ultralytics, pretrained checkpoint as the starting point) on the
boxes built in notebook 02, evaluated on the held-out **test** split from
`data/split.json` -- the same split notebook 03's U-Net is evaluated on.

**Meant to run on a free GPU notebook (Colab/Kaggle)** with the full downloaded
dataset (`--n 300`+) and a real epoch count (`EPOCHS = 50` is a reasonable start).
The output cells committed to this notebook are instead a **smoke test** -- a handful
of epochs on the 20-chip sample this repo verified everything against -- proving the
training code itself runs correctly end to end. Its metrics are **not meant to be
real results**; retrain properly before reading anything into the numbers.

```python
import sys
sys.path.insert(0, '..')

import json

from ultralytics import YOLO

# Set EPOCHS = 50+ for a real run on a GPU. This notebook's own committed
# output used a small value -- see the markdown note above.
EPOCHS = 15
IMGSZ = 256
```

## Train

`yolo11n.pt` is the smallest pretrained YOLO11 checkpoint -- transfer learning, not
training from scratch. `data/yolo/data.yaml`'s `train`/`val` keys already point at
notebook 02's per-split `train.txt`/`val.txt`.

```python
model = YOLO('yolo11n.pt')

results = model.train(
    data='../data/yolo/data.yaml',
    epochs=EPOCHS,
    imgsz=IMGSZ,
    project='../runs',
    name='yolo',
    exist_ok=True,
    verbose=False,
    plots=False,
)
```

## Evaluate on the held-out test split

`split='test'` tells Ultralytics to evaluate against the `test` key in `data.yaml`
(`test.txt`) rather than the `val` split it already saw during training.

```python
test_metrics = model.val(data='../data/yolo/data.yaml', split='test', imgsz=IMGSZ, plots=False)

test_map50 = float(test_metrics.box.map50)
test_map50_95 = float(test_metrics.box.map)
print(f'test mAP@0.5:      {test_map50:.3f}')
print(f'test mAP@0.5:0.95: {test_map50_95:.3f}')
```

## Save metrics (weights are already saved under `../runs/yolo/weights/` by `model.train`)

```python
with open('../runs/yolo_metrics.json', 'w') as f:
    json.dump({'test_map50': test_map50, 'test_map50_95': test_map50_95, 'epochs': EPOCHS}, f, indent=2)

print('saved ../runs/yolo/weights/best.pt and ../runs/yolo_metrics.json')
```
